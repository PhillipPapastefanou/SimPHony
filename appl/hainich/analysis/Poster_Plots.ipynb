```python
import pandas as pd
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import os

root_output_directory = "/Users/pp/data/Simulations/A08_Hydraulics_standalone/hainich"
print(os.listdir(root_output_directory))
```

```python
scenario_name = "fine_vg_more_constrained"
input_path = os.path.join(root_output_directory, scenario_name, 'input')
output_path = os.path.join(root_output_directory, scenario_name, 'output')
post_path = os.path.join(root_output_directory, scenario_name, 'post')

best_J = pd.read_csv(os.path.join(post_path, 'parameters_best_J.csv'))
best_psi_stem = pd.read_csv(os.path.join(post_path, 'parameters_best_psi_stem.csv'))
best_both = pd.read_csv(os.path.join(post_path, 'parameters_best_both.csv'))
```

```python
max_rmse_psi_stem = 0.075
max_rmse_J = 0.00012

df_both_sel = best_both[(best_both['rmse_psi_stem'] < max_rmse_psi_stem) & (best_both['rmse_J'] < max_rmse_J)].sort_values('rmse_com_avg')
df_J_sel = best_J[best_J['rmse_J'] < max_rmse_J].sort_values('rmse_J')
df_psi_stem_sel = best_psi_stem[best_psi_stem['rmse_psi_stem'] < max_rmse_psi_stem].sort_values('rmse_psi_stem')

print(best_both[(best_both['rmse_psi_stem'] < max_rmse_psi_stem) & (best_both['rmse_J'] < max_rmse_J)].shape[0])
print(best_psi_stem[best_psi_stem['rmse_psi_stem'] < max_rmse_psi_stem].shape[0])
print(best_J[best_J['rmse_J'] < max_rmse_J].shape[0])


df_both_sel.to_csv(f"{post_path}/df_both_sel_ordered.csv")
df_J_sel.to_csv(f"{post_path}/df_J_sel_ordered.csv")
df_psi_stem_sel.to_csv(f"{post_path}/df_psi_stem_sel_ordered.csv")
```

```python
vars = ['psi_leaf_50_close', 'd_50_close', 'stem_hydraulic_capacitance', 'k_xylem_sat', 'huber_value',
        'leaf_hydraulic_capacitance', 'psi50_xylem', 'leaf_area_index', 'g0', 'g1', 'k_soil_sat0', 'psi_soil_sat0',
        'jackson_root_beta', 'theta_s0', 'pore_0', 'theta_r0', 'anet_max', 'root_area_index', 'tree_density', 'g_bark']

for var in vars:

    if var == 'leaf_hydraulic_capacitance':
        plt.hist(np.log10(best_psi_stem[var]), alpha=0.5, bins=20, density=True);
        plt.hist(np.log10(best_J[var]), alpha=0.5, bins=20, density=True);
        plt.hist(np.log10(df_both_sel[var]), alpha=0.5, bins=20, density=True);
    elif var == 'k_soil_sat0':
        plt.hist(np.log10(best_psi_stem[var]), alpha=0.5, bins=20, density=True);
        plt.hist(np.log10(best_J[var]), alpha=0.5, bins=20, density=True);
        plt.hist(np.log10(df_both_sel[var]), alpha=0.5, bins=20, density=True);
    else:
        plt.hist(best_psi_stem[var], alpha=0.5, bins=20, label='stem', density=True);
        plt.hist(best_J[var], alpha=0.5, bins=20, label='J', density=True);
        plt.hist(df_both_sel[var], alpha=0.5, bins=20, label='both', density=True);

    plt.legend()
    plt.title(var)
    plt.savefig(f"{post_path}/p_hist_{var}.png")
    plt.close()
```
