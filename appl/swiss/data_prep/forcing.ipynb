```python
dm =pd.merge(depth_list_df[0]['q50'],df2['mean soil water content (m3/m3)'], left_index=True, right_index=True)

dm.rename(columns={ 'q50': "q50_10" }, inplace = True)
dm =pd.merge(depth_list_df[1]['q50'],dm, left_index=True, right_index=True)
dm.rename(columns={ 'q50': "q50_40" }, inplace = True)

dm =pd.merge(depth_list_df[2]['q50'], dm, left_index=True, right_index=True)
dm.rename(columns={ 'q50': "q50_80" }, inplace = True)
dm
```

```python
dm['diff10'] = dm['q50_10'] / 100.0 - dm['mean soil water content (m3/m3)']
```

```python
dm['v10_adapted']= dm['q50_10']/100 - dm['diff10']
dm['v40_adapted']= dm['q50_40']/100 - dm['diff10']
dm['v80_adapted']= dm['q50_80']/100 - dm['diff10']
```

```python
plt.plot(dm['v10_adapted'], color='red', label='10cm depth')
plt.plot(dm['v40_adapted'], color='blue', label='40cm depth')
plt.plot(dm['v80_adapted'], color='green', label='80cm depth')
```

```python
dm['sd'] = df2['sd']
dm.index.name = 'date'
```
