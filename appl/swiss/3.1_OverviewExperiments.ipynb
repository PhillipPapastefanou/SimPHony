```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
scenarions = {}
data_alive= {}
data_alive_a= {}
data_dead= {}
```

```python
#scenarions['all'] = 'cond_alivegrp_12_var_psiL_full_psiX_4544510'
scenarions['var Cs\nvar LAI'] = 'cond_alivegrp_12_var_psiL_full_psiX_4544510'
scenarions['low const Cs\nvar LAI'] = 'cond_alivegrp_12_no_psi_stemL_4544510'
scenarions['low const Cs\nconst LAI'] = 'cond_alivegrp_12_no_psi_stemL_no_lai_4544510'
scenarions['high const Cs\nvar LAI'] = 'cond_alivegrp_12_no_psi_stem_H_4544510'
scenarions['high const Cs\nno LAI'] = 'cond_alivegrp_12_no_psi_stem_H_no_lai_4544510'

range_dead = [0, 3, 6, 7]
```

```python
for name in scenarions:
    folder = scenarions[name]
    file = f"{folder}/PR_params_all_alive_rmse.csv"
    data_alive[name] = pd.read_csv(file)['rmse'].iloc[0:1000]/10.0
    data_alive_a[name] = pd.read_csv(file).iloc[0:1000]

    data_dead_tree = {}
    for i in range_dead:
        file = f"{folder}/PR_parameters_{i}_rmse.csv"
        data_dead_tree[i] =pd.read_csv(file)['rmse'].iloc[0:1000]

    data_dead[name] = data_dead_tree
```

```python
fig = plt.figure(figsize=(8,12))
ax = fig.add_subplot(3,2,1);
ax.boxplot([data_alive[name] for name in scenarions])
#ax.set_ylim(0.4, 1.3)
ax.set_xticklabels(labels= [name for name in scenarions] , rotation=60)
s = 2
for i in range_dead:
    ax = fig.add_subplot(3,2,s);
    ax.boxplot([data_dead[name][i] for name in scenarions], labels= [name for name in scenarions] )
    s+=1
    #ax.set_ylim(0.4, 1.3)
    ax.set_xticklabels(labels= [name for name in scenarions] , rotation=60)
plt.subplots_adjust(hspace = 0.5)
plt.show()
```
